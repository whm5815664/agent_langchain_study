## 1.2 LangChain 多模态消息构造

#### 1. 使用 `encode_image` 编码本地图片

多模态模型通常不能直接读取本地图片路径，因此需要先读取图片的二进制数据，并将其转换为 Base64 字符串。


#### 2. 使用 `HumanMessage` 封装多模态消息

LangChain 使用 `HumanMessage` 表示用户消息。多模态消息的 `content` 参数是一个列表，可以同时放入文本块和图片块。



````python
import os
from langchain.messages import HumanMessage

# 压缩并编码图片
def encode_image(path, max_kb=300):
    from io import BytesIO
    from PIL import Image
    import base64

    image = Image.open(path).convert("RGB")
    while True:
        for quality in range(90, 10, -5):
            buffer = BytesIO()
            image.save(buffer, "JPEG", quality=quality, optimize=True)

            if buffer.tell() <= max_kb * 1024:
                return base64.b64encode(buffer.getvalue()).decode()

        image.thumbnail((int(image.width * 0.8), int(image.height * 0.8)))


#下载网络图片
def download_image(url, save_path=None):
    # 请求图片
    response = requests.get(
        url,
        timeout=30,
        verify=False,
        headers={
            "User-Agent": "Mozilla/5.0"
        }
    )
    response.raise_for_status()
    # 如果没有指定保存路径，则根据URL自动生成文件名
    if save_path is None:
        filename = os.path.basename(urlparse(url).path)
        # URL没有文件名时
        if not filename:
            filename = "download_image.jpg"
        save_path = filename
    # 创建目录
    folder = os.path.dirname(save_path)
    if folder and not os.path.exists(folder):
        os.makedirs(folder)
    # 写入文件
    with open(save_path, "wb") as f:
        f.write(response.content)
    return save_path


# 构造消息
def build_msg(text=None, image_path=None):
    content = []
    # 添加文本
    if text:
        content.append({
            "type": "text",
            "text": text
        })

    # 添加图片
    if image_path:
        image_data = encode_image(image_path)

        content.append({
            "type": "image_url",
            "image_url": {
                "url": f"data:image/jpeg;base64,{image_data}",
                "detail": "auto"
            }
        })

    return [
        HumanMessage(content=content)
    ]
````

## 1.3 输出函数

本质是：agent对象.输出方法(message构造, config=记忆)

| 参数        | 类型                        | 是否必需 | 默认值    | 说明                                                                                            |
| --------- | ------------------------- | ---: | ------ | --------------------------------------------------------------------------------------------- |
| `agent`   | LangGraph Agent 或编译后的图对象  |    是 | 无      | 用于执行流式推理的智能体对象，需要支持 `stream()` 方法，例如通过 `create_react_agent()` 或 `StateGraph.compile()` 创建的对象。 |
| `message` | `list[BaseMessage]`       |    是 | 无      | 输入给智能体的消息列表，可包含 `HumanMessage`、`SystemMessage` 等消息对象。函数内部通过 `{"messages": message}` 传入智能体状态。  |
| `config`  | `dict` 或 `RunnableConfig` |    否 | `None` | 智能体运行配置，可用于传入线程编号、记忆等信息。                                                             |

````python
# 流式输入
def chat_stream(agent, message, config=None):
    result = agent.stream(
        {"messages":message}, 
        stream_mode='updates',
        config=config
    )
    for chunk in result:
        for node_name, update in chunk.items():
            
            # 新增空判断
            if not update:
                continue
            
            if "messages" in update:
                for msg in update["messages"]:
                    
                    # system提示词
                    if msg.type == "system":
                       print(f"[{msg.type}] {msg.content}")  
                   
                    # user文本内容
                    if msg.type == "human":
                       print(f"[{msg.type}] {msg.content}")  
                    
                    # ai回复
                    if msg.type == "ai":
                        if msg.content:
                            print(f"[{msg.type}] {msg.content}")
                        
                        if hasattr(msg, 'tool_calls') and msg.tool_calls:
                            calls = [tc['name'] for tc in msg.tool_calls]
                            print(f"请求调用: {calls}")
    
                    # tool工具回复
                    if msg.type == "tool":
                        print(f"[调用工具 {msg.name}]")
                        print(f"[{msg.type}] {msg.content}") 
````

````python
# 一次性输出
def chat_invoke(agent, message ,config=None):
    result = agent.invoke(
        {"messages":message}, 
        config=config
    )

    for msg in result["messages"]:
        # system提示词
        if msg.type == "system":
           print(f"[{msg.type}] {msg.content}")  
        
        # user文本内容
        if msg.type == "human":
           print(f"[{msg.type}] {msg.content}")  
        
        # ai文本内容
        if msg.type == "ai":
            print(f"[{msg.type}] {msg.content}") 
            
        # tool工具调用内容
        if msg.type == "tool":
            print(f"[调用工具 {msg.name}]")
            print(f"[{msg.type}] {msg.content}") 
````

# 2. 定义工具函数 @tool

````python
from langchain.tools import tool
import requests

@tool
def get_weather(longitude: str, latitude: str):
    '''
    功能:使用Open-Meteo查询指定坐标的天气情况。
    Args:
        longitude:经度，例如 "52.520011"
        latitude:纬度，例如 "13.410004"
    return:
        指定坐标未来逐小时的温度和相对湿度摘要。
    '''
    url = "https://api.open-meteo.com/v1/forecast"
    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": "temperature_2m,relative_humidity_2m",
        "timezone": "auto",
        "forecast_days": 1
    }
    try:
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()
        data = response.json()
        return data
    except requests.exceptions.RequestException as e:
        return f"天气接口请求失败：{str(e)}"


@tool
def get_calculate(expr: str):
    '''
    功能:根据用户输入的字符串表达式进行数学计算。
    Args:
        expr:表达式，例如 "1+1*1"
    return:
        计算的结果
    '''
    result = eval(expr)
    return result
         
````

````python
data = get_weather.invoke({
    "latitude": "52.520011",
    "longitude": "13.410004"
})
print(data)
````

# 4. 创建deepagent


### create_agent与create_deep_agent对比
| 功能     | `create_agent` 参数 | 调用方式                         | `create_deep_agent` 参数       | 调用方式                          | 区别                 |
| ------ | ----------------- | ---------------------------- | ---------------------------- | ----------------------------- | ------------------ |
| 模型     | `model`           | `model=llm`                  | `model`                      | `model=llm`                   | 相同                 |
| 工具     | `tools`           | `tools=[tool1,tool2]`        | `tools`                      | `tools=[tool1,tool2]`         | Deep Agent额外拥有内置工具 |
| 系统提示   | `system_prompt`   | `system_prompt="..."`        | `system_prompt`              | `system_prompt="..."`         | 相同                 |
| 中间件    | `middleware`      | `middleware=[xxx]`           | `middleware`                 | `middleware=[xxx]`            | 相同                 |
| 结构化输出  | `response_format` | `response_format=Schema`     | `response_format`            | `response_format=Schema`      | 相同                 |
| 状态管理   | `state_schema`    | `state_schema=MyState`       | 内部管理                         | 自动                            | Deep Agent封装       |
| 上下文    | `context_schema`  | `context_schema=Context`     | 内部管理                         | 自动                            | Deep Agent封装       |
| 记忆    | `checkpointer`    | `checkpointer=InMemorySaver()` | 支持                           | `checkpointer=...`            | 长任务恢复              |
| 存储     | `store`           | `store=store`                | 支持                           | `store=store`                 | 长期记忆               |
| 调试     | `debug`           | `debug=True`                 | 支持                           | `debug=True`                  | 相同                 |
| 名称     | `name`            | `name="agent"`               | 支持                           | `name="agent"`                | 相同                 |
| 子Agent | ❌                 | 无                            | `subagents`                  | `subagents=[...]`             | Deep Agent独有       |
| 技能     | ❌                 | 无                            | `skills`                     | `skills=["./skills"]`         | Claude Skill支持     |
| 文件系统   | ❌                 | 无                            | `backend`                    | `backend=FilesystemBackend()` | Deep Agent核心       |
| 权限控制   | ❌                 | 无                            | `interrupt_on` / permissions | 配置权限                          | 安全执行               |



````python
from deepagents import create_deep_agent

# 创建agent
agent = create_deep_agent(
    model=llm,
    tools=[get_weather, get_calculate],
    checkpointer=checkpointer
)
````

# 6. 状态控制

1） 获取当前对话状态

```python
state = agent.get_state(config)
```

2）获取消息历史：

```python
messages = state.values.get("messages", [])
```

3） 输出全部对话历史

```python
for msg in messages:
    print(f"[{msg.type}] {str(msg.content)[:500]}")
```

| `msg.type` | 说明     |
| ---------- | ------ |
| `system`   | 系统提示词  |
| `human`    | 用户消息   |
| `ai`       | 模型回复   |
| `tool`     | 工具执行结果 |


4） 统计历史模型调用 Token（提取单条 AI 消息的）
Token 用量通常记录在 `AIMessage` 的 `usage_metadata` 中：

```python
if msg.type == "ai":
    usage = getattr(msg, "usage_metadata", None)
    if usage:
        input_tokens = usage.get("input_tokens", 0) or 0
        output_tokens = usage.get("output_tokens", 0) or 0
        message_total = usage.get("total_tokens",input_tokens + output_tokens) or 0
```


````python
# 获取当前对话状态
state = agent.get_state(config)
messages = state.values.get("messages", [])

total_input_tokens = 0
total_output_tokens = 0
total_tokens = 0
counted_calls = 0

for index, msg in enumerate(messages):
    
    # 输出所有的对话
    print(f"  [{msg.type}] {str(msg.content)[:500]}")
    
    # Token 用量通常记录在 AIMessage 中
    if msg.type == "ai":
        usage = getattr(msg, "usage_metadata", None)
    
        if usage:
            input_tokens = usage.get("input_tokens", 0) or 0
            output_tokens = usage.get("output_tokens", 0) or 0
            message_total = usage.get("total_tokens", input_tokens + output_tokens) or 0
    
            total_input_tokens += input_tokens
            total_output_tokens += output_tokens
            total_tokens += message_total
            counted_calls += 1
    
            print(
                f"--------------------------"
                f"第 {index} 条 AI 消息："
                f"输入={input_tokens}，"
                f"输出={output_tokens}，"
                f"合计={message_total}"
                f"---------------------------\n"
            )

print("\n历史调用统计")
print(f"模型调用数：{counted_calls}")
print(f"累计输入 Token：{total_input_tokens}")
print(f"累计输出 Token：{total_output_tokens}")
print(f"累计 Token：{total_tokens}")
````

# Deep Agent + Claude Skills 子 Agent

通用执行任意 `./skills/*/SKILL.md`：读说明书、跑脚本、写代码生成文件。  
产物目录由 `OUTPUT_DIR` 指定。

```
主 Agent --task--> skills-agent（skills + Runtime 工具）
```

````python
# pip install deepagents pyyaml langchain-openai
import json
import os
import re
import shlex
import subprocess
from pathlib import Path

import yaml
from deepagents import create_deep_agent
from deepagents.backends import LocalShellBackend
from langchain.messages import HumanMessage
from langchain.tools import tool
from langchain_openai import ChatOpenAI
from langgraph.checkpoint.memory import InMemorySaver

# ========== 可改配置 ==========
ROOT = Path.cwd().resolve()
SKILLS_DIR = ROOT / "skills"
OUTPUT_DIR = ROOT / "outputs"          # 生成文件放这里（可改成任意目录）
SKILLS_VIRTUAL = "/skills/"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

llm = ChatOpenAI(
    model=os.getenv("LLM_MODEL", "qwen3.5-plus"),
    openai_api_key=os.getenv(
        "OPENAI_API_KEY",
        "sk-bvOWJIfQ8Y8IrZhzzpQZ80zSKHjWIHCkPXEgQ0H4Li5UCpUC",
    ),
    openai_api_base=os.getenv("OPENAI_API_BASE", "http://ai.wenmodel.com/v1"),
    temperature=0.3,
    max_tokens=4096,
)

backend = LocalShellBackend(
    root_dir=str(ROOT),
    virtual_mode=True,
    inherit_env=True,
    timeout=300,
)
print("skills:", [p.name for p in SKILLS_DIR.iterdir() if (p / "SKILL.md").exists()])
print("output:", OUTPUT_DIR)
````

## Skills Runtime（通用）

````python
SCRIPT_EXT = {".py", ".js", ".mjs", ".ts", ".sh", ".bat", ".ps1"}


def _parse_frontmatter(text: str) -> dict:
    m = re.match(r"^---\s*\n(.*?)\n---", text, re.DOTALL)
    if not m:
        return {}
    try:
        return yaml.safe_load(m.group(1)) or {}
    except Exception:
        return {}


def _load_skills(skills_dir: Path) -> dict[str, dict]:
    skills = {}
    for path in sorted(skills_dir.iterdir()):
        md = path / "SKILL.md"
        if not path.is_dir() or not md.exists():
            continue
        text = md.read_text(encoding="utf-8")
        meta = _parse_frontmatter(text)
        name = str(meta.get("name") or path.name)
        scripts = [
            f.relative_to(path).as_posix()
            for f in sorted(path.rglob("*"))
            if f.is_file()
            and f.suffix.lower() in SCRIPT_EXT
            and not any(p in {"node_modules", ".git", "__pycache__", "schemas"} for p in f.parts)
        ]
        skills[name] = {
            "name": name,
            "description": str(meta.get("description") or "").strip(),
            "path": path,
            "content": text,
            "scripts": scripts,
        }
    return skills


def _run(cmd: list[str], cwd: Path) -> dict:
    try:
        r = subprocess.run(
            cmd,
            cwd=str(cwd),
            capture_output=True,
            text=True,
            timeout=300,
            env={**os.environ, "OUTPUT_DIR": str(OUTPUT_DIR), "SKILLS_DIR": str(SKILLS_DIR)},
        )
        return {
            "ok": r.returncode == 0,
            "returncode": r.returncode,
            "stdout": (r.stdout or "")[-6000:],
            "stderr": (r.stderr or "")[-6000:],
            "cmd": cmd,
            "cwd": str(cwd),
        }
    except Exception as e:
        return {"ok": False, "error": str(e), "cmd": cmd}


def _cmd_for(script: Path, argv: list[str]) -> list[str]:
    s = script.suffix.lower()
    if s == ".py":
        return ["python", str(script), *argv]
    if s in {".js", ".mjs"}:
        return ["node", str(script), *argv]
    if s == ".ts":
        return ["npx", "--yes", "tsx", str(script), *argv]
    if s == ".ps1":
        return ["powershell", "-ExecutionPolicy", "Bypass", "-File", str(script), *argv]
    if s == ".bat":
        return [str(script), *argv]
    if s == ".sh":
        return ["bash", str(script), *argv]
    raise ValueError(f"不支持: {s}")


SKILLS = _load_skills(SKILLS_DIR)
print("loaded:", list(SKILLS))


@tool
def list_skills() -> str:
    """列出所有可用 Claude Skills。"""
    data = [
        {"name": s["name"], "description": s["description"], "scripts": s["scripts"]}
        for s in SKILLS.values()
    ]
    return json.dumps(data, ensure_ascii=False, indent=2)


@tool
def load_skill(skill_name: str) -> str:
    """加载 skill 完整 SKILL.md 与脚本列表。执行前必做。"""
    if skill_name not in SKILLS:
        return json.dumps({"ok": False, "error": f"不存在: {skill_name}", "可用": list(SKILLS)}, ensure_ascii=False)
    s = SKILLS[skill_name]
    return json.dumps(
        {
            "name": s["name"],
            "description": s["description"],
            "scripts": s["scripts"],
            "output_dir": str(OUTPUT_DIR),
            "hint": "按 content 用 run_skill_script / run_code 执行；产物必须写到 output_dir",
            "content": s["content"],
        },
        ensure_ascii=False,
        indent=2,
    )


@tool
def run_skill_script(skill_name: str, script_relpath: str, script_args: str = "") -> str:
    """执行 skill 内脚本。script_relpath 如 scripts/office/validate.py；script_args 为空格分隔参数。"""
    if skill_name not in SKILLS:
        return json.dumps({"ok": False, "error": f"不存在: {skill_name}"}, ensure_ascii=False)
    skill_path = SKILLS[skill_name]["path"]
    script = (skill_path / script_relpath).resolve()
    if not str(script).startswith(str(skill_path.resolve())) or not script.is_file():
        return json.dumps({"ok": False, "error": f"脚本无效: {script_relpath}"}, ensure_ascii=False)
    argv = shlex.split(script_args, posix=os.name != "nt") if script_args.strip() else []
    return json.dumps(_run(_cmd_for(script, argv), OUTPUT_DIR), ensure_ascii=False, indent=2)


@tool
def run_code(language: str, code: str, filename: str = "") -> str:
    """在 OUTPUT_DIR 写入并执行代码。language=python|javascript|js；文件请写到当前工作目录（即 OUTPUT_DIR）。"""
    lang = language.lower().strip()
    if lang in {"python", "py"}:
        name = filename or "_tmp_skill.py"
        runner = ["python"]
    elif lang in {"javascript", "js", "node"}:
        name = filename or "_tmp_skill.js"
        runner = ["node"]
    else:
        return json.dumps({"ok": False, "error": "language 仅支持 python / javascript"}, ensure_ascii=False)
    path = (OUTPUT_DIR / Path(name).name).resolve()
    path.write_text(code, encoding="utf-8")
    result = _run([*runner, str(path)], OUTPUT_DIR)
    result["file"] = str(path)
    return json.dumps(result, ensure_ascii=False, indent=2)


@tool
def list_output_files() -> str:
    """列出 OUTPUT_DIR 下的文件（任务结束前用来确认产物）。"""
    files = []
    for f in sorted(OUTPUT_DIR.rglob("*")):
        if f.is_file() and f.name not in {"_tmp_skill.py", "_tmp_skill.js"}:
            files.append(
                {
                    "path": str(f.relative_to(OUTPUT_DIR)).replace("\\", "/"),
                    "size": f.stat().st_size,
                }
            )
    return json.dumps({"output_dir": str(OUTPUT_DIR), "files": files}, ensure_ascii=False, indent=2)


skill_tools = [list_skills, load_skill, run_skill_script, run_code, list_output_files]
print("tools:", [t.name for t in skill_tools])
````

## skills-agent 子 Agent + 主 Agent

````python
SKILLS_PROMPT = f"""你是 Claude Skills 执行专家（skills-agent）。

工作目录 / 产物目录固定为：{OUTPUT_DIR}
所有生成文件必须写到该目录（相对路径即可，cwd 已是该目录）。

流程：
1. list_skills 选 skill
2. load_skill 读完整说明书
3. 按说明书执行：run_skill_script 或 run_code（python/javascript）
4. list_output_files 确认产物

SKILL.md 不会自动执行；必须调用工具。需要产物却未生成 = 失败。
"""

skills_subagent = {
    "name": "skills-agent",
    "description": (
        "执行任意 Claude Skill：加载 SKILL.md、运行 scripts、按指南写代码并生成文件到指定输出目录。"
        "创建/编辑文档、跑 skill 脚本、按 skill 指南产出文件时委派给它。"
    ),
    "system_prompt": SKILLS_PROMPT,
    "skills": [SKILLS_VIRTUAL],
    "tools": skill_tools,
}

# 独立调试用
skills_agent = create_deep_agent(
    model=llm,
    backend=backend,
    tools=skill_tools,
    skills=[SKILLS_VIRTUAL],
    system_prompt=SKILLS_PROMPT,
    name="skills-agent-standalone",
)

checkpointer = InMemorySaver()
agent = create_deep_agent(
    model=llm,
    backend=backend,
    subagents=[skills_subagent],
    checkpointer=checkpointer,
    system_prompt=(
        "你是智能助手。涉及 Claude Skills / 文档生成 / 执行 skill 脚本时，"
        "必须用 task 委派给 skills-agent，并说明输出目录与文件名。普通问答可自己答。"
    ),
    name="main-agent",
)


def run_skills_agent(query: str) -> str:
    result = skills_agent.invoke({"messages": [HumanMessage(content=query)]})
    return result["messages"][-1].content


def chat(text: str, thread_id: str = "skills-demo") -> str:
    result = agent.invoke(
        {"messages": [HumanMessage(content=text)]},
        config={"configurable": {"thread_id": thread_id}},
    )
    return result["messages"][-1].content


print("ready: skills-agent + main-agent, OUTPUT_DIR =", OUTPUT_DIR)
````

## 演示

改 `OUTPUT_DIR` 后重新跑上面单元格即可切换输出目录。

````python
# 1) 直接调工具（无 LLM）
print(list_skills.invoke({}))
print(run_code.invoke({
    "language": "python",
    "code": (
        "from pathlib import Path\n"
        "Path('hello_from_skill.txt').write_text('ok from OUTPUT_DIR', encoding='utf-8')\n"
        "print('wrote', Path('hello_from_skill.txt').resolve())\n"
    ),
}))
print(list_output_files.invoke({}))
````

````python
# 2) 主 Agent 委派 skills-agent（通用 skill，产物进 OUTPUT_DIR）
print(chat(
    f"请通过 skills-agent，使用合适的 skill，在输出目录 {OUTPUT_DIR} "
    "生成一份 Word 文档 create_deep_agent_params.docx："
    "介绍 create_deep_agent 的参数 model/tools/system_prompt/subagents/skills/backend/checkpointer。"
    "按 SKILL.md 用 run_code 或 run_skill_script 完成，最后 list_output_files 确认。",
    thread_id="skills-demo-1",
))
print("\nOUTPUT_DIR 文件:", list_output_files.invoke({}))
````

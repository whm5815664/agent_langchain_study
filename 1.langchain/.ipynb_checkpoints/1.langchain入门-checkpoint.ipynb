# 2.创建agent

步骤:   
（1）用 @tool 装饰器把 Python 函数变成工具  
（2）用 create_agent() 创建能自动调用工具的 Agent  
（3）用 agent.invoke() 运行 Agent 并获得结果  



## 2.1 定义工具函数 @tool

用 @tool 装饰器定义一个工具  
函数的文档字符串就是工具的描述,模型会读取工具的描述来决定是否调用这个工具以及传什么参数。描述越清晰，模型就越不容易用错。


---
对于任意的python函数，只需加@tool修饰符，即可转换为agent可识别的工具函数，推荐格式为：
```python
@tool
def 函数名(参数1：类型, 参数2：类型):
    '''
        功能：描述
        Args：
            参数1：描述，例如
            参数2：描述，例如
        return：
            描述
    '''
    具体的python实现代码
```
---

````python
from langchain.tools import tool
import requests

@tool
def get_weather(longitude: str, latitude: str):
    '''
    功能:使用Open-Meteo查询指定坐标的天气情况。
    Args:
        longitude:经度，例如 "52.520011"
        latitude:纬度，例如 "13.410004"
    return:
        指定坐标未来逐小时的温度和相对湿度摘要。
    '''
    url = "https://api.open-meteo.com/v1/forecast"
    params = {
        "latitude": latitude,
        "longitude": longitude,
        "hourly": "temperature_2m,relative_humidity_2m",
        "timezone": "auto",
        "forecast_days": 1
    }
    try:
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()
        data = response.json()
        return data
    except requests.exceptions.RequestException as e:
        return f"天气接口请求失败：{str(e)}"


@tool
def get_calculate(expr: str):
    '''
    功能:根据用户输入的字符串表达式进行数学计算。
    Args:
        expr:表达式，例如 "1+1*1"
    return:
        计算的结果
    '''
    result = eval(expr)
    return result
    
        
    
````

````python
data = get_weather.invoke({
    "latitude": "52.520011",
    "longitude": "13.410004"
})
print(data)
````

## 2.2 创建智能体 create_agent(model, tools, system_prompt)

| 参数名              | 类型                                               | 描述                     |
|---------------------|----------------------------------------------------|--------------------------|
| `model`             | `str` \| `BaseChatModel`                           | 语言模型                 |
| `tools`             | `Sequence`                                         | 工具列表                 |
| `system_prompt`     | `str` \| `SystemMessage`                           | 系统提示                 |
| `middleware`        | `Sequence[AgentMiddleware]`                        | 中间件列表               |
| `response_format`   | `ResponseFormat` \| `type`                         | 结构化输出配置           |
| `state_schema`      | `type[AgentState]`                                 | 自定义状态结构           |
| `context_schema`    | `type`                                             | 运行时上下文结构         |
| `checkpointer`      | `Checkpointer`                                     | 对话持久化               |
| `store`             | `BaseStore`                                        | 跨会话存储               |
| `interrupt_before`  | `list[str]`                                        | 在哪些节点前暂停         |
| `interrupt_after`   | `list[str]`                                        | 在哪些节点后暂停         |
| `debug`             | `bool`                                             | 是否输出详细日志         |
| `name`              | `str`                                              | Agent 名称               |
| `cache`             | `BaseCache`                                        | 缓存配置                 |


````python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model

# 初始化智能体
agent = create_agent(
    model=llm,
    tools=[get_weather, get_calculate],
    system_prompt="你是一个aiot智能体。"
)
````

## 2.3 运行智能体 agent.invoke


````python
from langchain.messages import HumanMessage, AIMessage, SystemMessage

# 1. 构建输入消息
inputs = {
    "messages": [
        SystemMessage(content="你是一个天气助手，回答要简洁。"),   # prompt,给模型设置身份、规则、回答风格
        AIMessage(content="我可以帮你查询天气。"),           # 表示模型之前说过的话(一般用于保存历史对话，让模型知道前面自己回答过什么)
        HumanMessage(content="今天武汉天气怎么样？计算武汉最高温度-贵阳最高温度")              # 用户输入
    ]
}


# 2. invoke() 运行 Agent，返回最终状态
result = agent.invoke(inputs)
print('响应状态:\n', result)
print('\n--------------\n')
print('响应消息:\n', result['messages'])
print('\n--------------\n')


# 3. 查看消息历史（包含 AI 的工具调用和工具返回结果）
print("=== 完整消息历史 ===")
for msg in result["messages"]:
    # system提示词
    if msg.type == "system":
       print(f"[{msg.type}] {msg.content}")  
    
    # user文本内容
    if msg.type == "human":
       print(f"[{msg.type}] {msg.content}")  
    
    # ai文本内容
    if msg.type == "ai":
        print(f"[{msg.type}] {msg.content}") 
        
    # tool工具调用内容
    if msg.type == "tool":
        print(f"[调用工具 {msg.name}]")
        print(f"[{msg.type}] {msg.content}") 

print('\n--------------\n')


# 4. 输出最终答案
print(result["messages"][-1].content)
````

# 3. LangChain 流式输出agent.stream()


````python
# 1.创建agent
agent = create_agent(
    model=llm,
    system_prompt="你是牛马",
    tools=[get_weather],
)
````

````python
# 构建用户输入
inputs = {
    "messages": [HumanMessage(content="今天武汉天气怎么样？")]
}
````

## 3.1. 使用stream_mode='messages' 逐 Token 流式

````python
result = agent.stream(inputs, stream_mode='messages')
print(result)
print('---------------\n')


for msg_chunk, metadata in result:
    if msg_chunk.content:        # 有消息采输出
        print(msg_chunk.content, end="", flush=True)

````

## 3.2 使用stream_mode="updates"逐步查看Agent执行过程
node_name含有一下内容：
| 节点名称   | 说明                                                                                           |
|-----------|------------------------------------------------------------------------------------------------|
| `"model"` | Agent 决策节点：LLM 在此决定下一步行动（调用工具或直接回复）。如果输出包含 `tool_calls`，说明要调用工具；否则为最终回复。 |
| `"tools"` | 工具执行节点：运行 Agent 选择的工具，并返回 `ToolMessage`。                                     |
                                

````python
result = agent.stream(inputs, stream_mode='updates')

for chunk in result:
    for node_name, update in chunk.items():
        print('-----------------------------')
        print(f"节点：{node_name}\n", end=" ")
        
        if "messages" in update:
            for msg in update["messages"]:
                
                # system提示词
                if msg.type == "system":
                   print(f"[{msg.type}] {msg.content}")  
               
                # user文本内容
                if msg.type == "human":
                   print(f"[{msg.type}] {msg.content}")  
                
                # ai回复
                if msg.type == "ai":
                    if msg.content:
                        print(f"[{msg.type}] {msg.content}")
                    
                    if hasattr(msg, 'tool_calls') and msg.tool_calls:
                        calls = [tc['name'] for tc in msg.tool_calls]
                        print(f"请求调用: {calls}")

                # tool工具回复
                if msg.type == "tool":
                    print(f"[调用工具 {msg.name}]")
                    print(f"[{msg.type}] {msg.content}") 

    
````

### 3.3（可选）封装调用函数

````python
def chat_stream(message):
    result = agent.stream(
        {"messages":message}, 
        stream_mode='updates'
    )

    # token统计
    total_tokens = 0
    total_input_tokens = 0
    total_output_tokens = 0
    
    for chunk in result:
        for node_name, update in chunk.items():
            print('-----------------------------')
            print(f"节点：{node_name}\n", end=" ")
            
            if "messages" in update:
                for msg in update["messages"]:
                    
                    # system提示词
                    if msg.type == "system":
                       print(f"[{msg.type}] {msg.content}")  
                   
                    # user文本内容
                    if msg.type == "human":
                       print(f"[{msg.type}] {msg.content}")  
                    
                    # ai回复
                    if msg.type == "ai":
                        if msg.content:
                            print(f"[{msg.type}] {msg.content}")
                        
                        if hasattr(msg, 'tool_calls') and msg.tool_calls:
                            calls = [tc['name'] for tc in msg.tool_calls]
                            print(f"请求调用: {calls}")
    
                    # tool工具回复
                    if msg.type == "tool":
                        print(f"[调用工具 {msg.name}]")
                        print(f"[{msg.type}] {msg.content}") 

                    # 计算token
                    if hasattr(msg, "usage_metadata") and msg.usage_metadata:
                        usage = msg.usage_metadata
                        total_tokens += usage.get("total_tokens", 0)
                        total_input_tokens += usage.get("input_tokens", 0)
                        total_output_tokens += usage.get("output_tokens", 0)

    
    print("\n=============================")
    print("Token统计:")
    print(f"输入 tokens: {total_input_tokens}")
    print(f"输出 tokens: {total_output_tokens}")
    print(f"总计 tokens: {total_tokens}")



msg = [
    AIMessage(content="lsq是卷王"),
    HumanMessage(content="lsq是谁？")
]

# 输入智能体
chat_stream(msg)
````

# 4. web异步流式SSE输出agent.astream()（做网页时使用）

````python
# 创建agent
agent = create_agent(
    model=llm,
    system_prompt="你是牛马",
    tools=[get_weather, get_calculate],
)

                
````

````python
async def chat(message: str):
    async def generate():
        async for msg_chunk, metadata in agent.astream(
            {"messages": [HumanMessage(content=message)]},
            stream_mode="messages",
        ):
            if msg_chunk.content:
                yield msg_chunk.content
                #yield f"data: {msg_chunk.content}\n\n"
        yield "[DONE]"
        #yield "data: [DONE]\n\n"

    return generate()

    # return StreamingResponse(
    #     generate(),
    #     media_type="text/event-stream",
    # )

gen = await chat("武汉天气如何")  
async for chunk in gen:
    print(chunk, end="", flush=True)
````

## 4.（可选） 含思维链的输出

````python
# 创建agent
agent = create_agent(
    model=llm,
    system_prompt="你是牛马",
    tools=[get_weather, get_calculate],
)



async def chat_with_cot(message: str):
    async def generate():
        async for chunk in agent.astream(
            {"messages": [HumanMessage(content=message)]},
            stream_mode="updates",
        ):
            for node_name, update in chunk.items():
                print('---------------------------')
                print(f"当前节点: {node_name} \n")    
                
                # agent决策节点
                if node_name == "model":
                    print("model决策节点\n")
                    for msg in update.get("messages", []):     
                        if msg.type == "ai":
                            if hasattr(msg, "tool_calls") and msg.tool_calls:
                                for tc in msg.tool_calls:
                                    yield f"[思考] 调用工具 {tc['name']}，参数 {tc.get('args')}\n"
                            if msg.content:
                                yield f"[回复] {msg.content}\n"

                                
                # 工具执行节点
                if node_name == "tools":
                    print("tools工具执行节点\n")
                    for msg in update.get("messages", []):
                        if msg.type == "tool":
                            yield f"[工具] {msg.name} 返回：{msg.content}\n"
                        if msg.type == "ai":
                            yield f"[ai] {msg.content}\n"


                
        yield "[DONE]\n"
    return generate()

# 运行
gen = await chat_with_cot("武汉今日最高温度-贵阳今日最高温度为多少")
async for line in gen:
    print(line, end="")
````
